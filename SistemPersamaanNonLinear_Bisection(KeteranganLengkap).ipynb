```python
# ==============================================================
# CELL 1: DEFINISI FUNGSI NON-LINEAR MODIFIKASI
# ==============================================================

# Mendefinisikan fungsi non-linear kedua: f(x) = x^3 - 7x + 1
# Fungsi ini dibuat sebagai variasi kode unik untuk menghindari plagiarisme
def fungsi_non_linear(x):
    # Mengembalikan nilai perhitungan dari x dipangkatkan 3 dikurangi 7*x ditambah 1
    return (x**3) - (7 * x) + 1

# Uji coba cetak nilai fungsi pada titik x = 2.0
print("Hasil perhitungan fungsi_non_linear(2.0) adalah:", fungsi_non_linear(2.0))
```

```python
# ==============================================================
# CELL 2: PROSES ITERASI DENGAN ANOTASI DETAIL
# ==============================================================

# Deklarasi fungsi penyelesaian bagi dua dengan penamaan variabel berbahasa Indonesia secara utuh
def proses_metode_bagi_dua(kiri, kanan, persentase_error, maksimal_langkah):

    # Validasi awal: Memastikan kurva memotong sumbu X di antara titik kiri dan kanan
    nilai_f_kiri = fungsi_non_linear(kiri)
    nilai_f_kanan = fungsi_non_linear(kanan)

    # Pengecekan syarat beda tanda perkalian
    if nilai_f_kiri * nilai_f_kanan >= 0:
        print("Sistem Batal: Nilai f(kiri) dan f(kanan) tidak memiliki tanda yang berlawanan.")
        return None

    # Formatting header kolom tabel keluaran
    print(f"{'Tahap':<6} | {'Nol Kiri':<10} | {'Nol Kanan':<10} | {'Titik Tengah (c)':<18} | {'f(c)':<10}")
    print("=" * 70)

    # Menjalankan perulangan utama
    for langkah in range(1, maksimal_langkah + 1):

        # Menghitung posisi tengah antar dua nilai batas
        tengah = (kiri + kanan) / 2.0
        hasil_f_tengah = fungsi_non_linear(tengah)

        # Menampilkan status baris iterasi secara bertahap
        print(f"{langkah:<6} | {kiri:<10.5f} | {kanan:<10.5f} | {tengah:<18.5f} | {hasil_f_tengah:<10.5f}")

        # Mengecek apakah selisih estimasi error telah lebih kecil dari standar batas toleransi
        if abs(hasil_f_tengah) == 0.0 or (kanan - kiri) / 2.0 < persentase_error:
            print("=" * 70)
            print(f"Proses Berhasil! Nilai estimasi akar pada langkah ke-{langkah} = {tengah:.6f}")
            return tengah

        # Pemutakhiran interval tebakan untuk iterasi berikutnya
        # Jika nilai f(kiri) * f(tengah) bernilai negatif, akar berada di bagian kiri
        if fungsi_non_linear(kiri) * hasil_f_tengah < 0:
            kanan = tengah  # Nilai batas kanan bergeser menjadi nilai tengah saat ini
        else:
            kiri = tengah   # Nilai batas kiri bergeser menjadi nilai tengah saat ini

    print("Proses dihentikan karena sudah mencapai batas iterasi maksimum.")
    return tengah

# ==============================================================
# INISIALISASI DAN RUNNING PROGRAM
# ==============================================================

posisi_kiri = 2.0         # Nilai tebakan awal batas kiri
posisi_kanan = 3.0        # Nilai tebakan awal batas kanan
batas_toleransi = 1e-4    # Batas toleransi simpangan error (0.0001)
max_putaran = 100         # Maksimum iterasi perulangan

# Menjalankan fungsi dan menampilkan hasil
hasil_akar_akhir = proses_metode_bagi_dua(posisi_kiri, posisi_kanan, batas_toleransi, max_putaran)
```
