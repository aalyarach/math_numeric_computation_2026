```python
# ==============================================================
# CELL 1: DEFINISI FUNGSI NON-LINEAR
# ==============================================================

# Mendefinisikan fungsi matematika f(x) yang akan dicari nilai akarnya
# Persamaan yang digunakan: f(x) = x^3 - x - 2
def f(x):
    # Operator ** digunakan di Python untuk melakukan operasi pemangkatan (x pangkat 3)
    # Mengembalikan hasil evaluasi nilai f(x) berdasarkan masukan parameter x
    return x**3 - x - 2

# Mengetes fungsi dengan nilai contoh x = 2 untuk memastikan fungsi berjalan
print("Uji coba fungsi f(2) =", f(2))
```

```python
# ==============================================================
# CELL 2: IMPLEMENTASI ALGORITMA BISECTION DAN TABEL ITERASI
# ==============================================================

# Mendefinisikan fungsi utama bisection yang menerima 4 parameter masukan:
# a = batas kiri interval awal, b = batas kanan interval awal
# tol = nilai toleransi batas error, max_iter = batas maksimum putaran perulangan
def bisection_detail(a, b, tol, max_iter):

    # Syarat utama Metode Bisection: f(a) dan f(b) harus berlawanan tanda (f(a) * f(b) < 0)
    # Jika f(a) * f(b) >= 0, artinya tidak dijamin ada akar di antara rentang a dan b
    if f(a) * f(b) >= 0:
        print("Kesalahan: Nilai f(a) dan f(b) harus memiliki tanda berlawanan!")
        return None

    # Mencetak bagian tata letak header tabel agar tampilan output rapi
    print(f"{'Iterasi':<8} | {'Batas a':<10} | {'Batas b':<10} | {'Titik c':<10} | {'Nilai f(c)':<10}")
    print("-" * 65)

    # Memulai perulangan iterasi dari 1 hingga nilai batas maksimum iterasi (max_iter)
    for i in range(1, max_iter + 1):

        # Mencari titik tengah (c) dengan membagi dua jumlah batas kiri (a) dan batas kanan (b)
        c = (a + b) / 2.0

        # Mencetak data hasil kalkulasi pada iterasi saat ini ke dalam format kolom tabel
        print(f"{i:<8} | {a:<10.4f} | {b:<10.4f} | {c:<10.4f} | {f(c):<10.4f}")

        # Kondisi penghentian (terminasi):
        # 1. Jika nilai f(c) persis bernilai 0.0 (akar eksak)
        # 2. Atau selisih lebar interval (b - a)/2 sudah lebih kecil dari toleransi error (tol)
        if f(c) == 0.0 or (b - a) / 2.0 < tol:
            print("-" * 65)
            print(f"Kondisi terpenuhi! Akar persamaan ditemukan pada iterasi ke-{i}: x = {c:.6f}")
            return c

        # Pembaharuan batas interval (Bracket updating):
        # Jika f(a) dikali f(c) kurang dari 0, berarti akar berada di rentang sebelah kiri [a, c]
        if f(a) * f(c) < 0:
            b = c  # Geser posisi batas kanan (b) ke titik tengah (c)
        # Jika tidak, berarti akar berada di rentang sebelah kanan [c, b]
        else:
            a = c  # Geser posisi batas kiri (a) ke titik tengah (c)

    print("Iterasi dihentikan karena sudah menyentuh batas maksimum perulangan.")
    return c

# ==============================================================
# EKSEKUSI PROGRAM DENGAN PARAMETER
# ==============================================================

# Menetapkan nilai tebakan awal dan parameter toleransi
batas_a = 1.0          # Batas awal interval kiri
batas_b = 2.0          # Batas awal interval kanan
toleransi_error = 1e-5 # Nilai toleransi simpangan error (0.00001)
limit_iterasi = 50     # Jumlah maksimal perulangan yang diizinkan

# Memanggil fungsi bisection_detail untuk me-running kalkulasi
akar_persamaan = bisection_detail(batas_a, batas_b, toleransi_error, limit_iterasi)
```
