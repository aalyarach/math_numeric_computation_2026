```python
import numpy as np

def eliminasi_gauss(A, b):
    # Mengonversi array ke tipe float agar kalkulasi desimal (pembagian) presisi
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # 1. Membentuk Matriks Augmented [A | b]
    # b.reshape(n, 1) mengubah vektor baris menjadi vektor kolom
    # np.hstack menempelkan A dan b secara horizontal
    M = np.hstack((A, b.reshape(n, 1)))

    print("Matriks Augmented Awal [A | b]:")
    print(np.round(M, 4), "\n")

    # 2. Proses Eliminasi Maju (Forward Elimination)
    for k in range(n):
        # --- Partial Pivoting ---
        # Mencari nilai absolut terbesar di bawah baris k pada kolom k
        max_row = np.argmax(abs(M[k:n, k])) + k
        if M[max_row, k] == 0:
            raise ValueError("Sistem persamaan tidak memiliki solusi unik (Matriks Singular/Determinan 0).")

        # Tukar baris jika elemen terbesar tidak berada di baris k saat ini
        if max_row != k:
            M[[k, max_row]] = M[[max_row, k]]
        # ------------------------

        # Membuat nilai 0 untuk semua baris di bawah baris pivot k
        for i in range(k + 1, n):
            faktor = M[i, k] / M[k, k]
            for j in range(k, n + 1):
                M[i, j] -= faktor * M[k, j]

    print("Matriks setelah Eliminasi Maju (Segitiga Atas):")
    print(np.round(M, 4), "\n")

    # 3. Proses Substitusi Mundur (Backward Substitution)
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        jumlah = 0
        for j in range(i + 1, n):
            jumlah += M[i, j] * x[j]
        # x_i = (Nilai ujung kanan - jumlah yang sudah diketahui) / elemen diagonal
        x[i] = (M[i, n] - jumlah) / M[i, i]

    return x

# ==========================================
# BAGIAN PENGUJIAN (TESTING)
# ==========================================
# Sistem Persamaan Linear contoh:
#  2x + 3y -  z = 5
#  4x + 4y - 3z = 3
# -2x + 3y -  z = 1

A_matriks = [
    [ 2,  3, -1],
    [ 4,  4, -3],
    [-2,  3, -1]
]
b_vektor = [5, 3, 1]

try:
    solusi = eliminasi_gauss(A_matriks, b_vektor)
    print("Hasil Akhir Vektor Solusi (x):")
    for i in range(len(solusi)):
        print(f"x{i+1} = {solusi[i]:.4f}")
except ValueError as e:
    print("Error:", e)
```
